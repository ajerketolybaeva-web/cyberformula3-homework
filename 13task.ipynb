```python
import csv
import json
import numpy as np

with open("orders_1.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

print(len(rows))
print(rows[0])
```

```python
# Берём названия первых пяти товаров
products = [rows[i]["product"] for i in range(5)]

# Записываем в файл
with open("products.txt", "w", encoding="utf-8") as f:
    for product in products:
        f.write(product + "\n")

# Читаем файл обратно
with open("products.txt", "r", encoding="utf-8") as f:
    products_read = [line.strip() for line in f if line.strip()]

print(products_read)
```

```python
a = np.arange(1, 25)

table = a.reshape(4, 6)

print(table)
print("Число:", table[2, 1])
print("Последний столбец:", table[:, -1])
print("Сумма первой строки:", table[0].sum())
print("Форма при 6 строках:", a.reshape(6, 4).shape)
```

```python
price = np.array([
    int(float(row["price"].replace("₸", "").replace(" ", "").strip()))
    for row in rows
])

qty = np.array([int(row["qty"]) for row in rows])

product = np.array([row["product"] for row in rows])

revenue = price * qty

# 1. Количество заказов
print("Всего заказов:", len(rows))

# 2. Самый дешёвый заказ
min_index = revenue.argmin()

print("Самый дешёвый:", revenue[min_index])
print("Товар:", product[min_index])

# 3. Самый дорогой заказ
max_index = revenue.argmax()

print("Самый дорогой:", revenue[max_index])
print("Во сколько раз дороже:", revenue[max_index] / revenue[min_index])
```

```python
report = {
    "заказов": int(len(rows)),
    "самый_дешёвый": {
        "сумма": int(revenue[min_index]),
        "товар": str(product[min_index])
    },
    "самый_дорогой": int(revenue[max_index])
}

# Сохраняем
with open("hw_base.json", "w", encoding="utf-8") as f:
    json.dump(report, f, ensure_ascii=False, indent=4)

# Читаем
with open("hw_base.json", "r", encoding="utf-8") as f:
    data = json.load(f)

print(data["самый_дешёвый"]["товар"])
```

```python
status = np.array([row["status"].strip().lower() for row in rows])

cancelled = status == "отменён"

print("Отменённых заказов:", cancelled.sum())
print("Доля отмен:", cancelled.mean() * 100, "%")
print("Потеря:", revenue[cancelled].sum(), "₸")
```

```python
week = revenue[:35]

daily_orders = week.reshape(7, 5)

daily_revenue = daily_orders.sum(axis=1)

print("Форма:", daily_orders.shape)
print("Выручка за каждый день:", daily_revenue)

best_day = daily_revenue.argmax()

print("Самый прибыльный день:", best_day + 1)
print("Выручка:", daily_revenue[best_day])
```

```python
city = np.array([row["city"].strip() for row in rows])

unique_cities = np.unique(city)

cities_report = {}

for c in unique_cities:
    mask = city == c

    cities_report[str(c)] = {
        "заказов": int(mask.sum()),
        "средний_чек": int(revenue[mask].mean())
    }

print(cities_report)

# Сохраняем в JSON
with open("cities.json", "w", encoding="utf-8") as f:
    json.dump(cities_report, f, ensure_ascii=False, indent=4)

# Находим город с самым высоким средним чеком
best_city = max(cities_report, key=lambda c: cities_report[c]["средний_чек"])

print("Город с самым высоким средним чеком:", best_city)
print("Средний чек:", cities_report[best_city]["средний_чек"])
```

```python
def clean_price(s):
    s = str(s).replace("₸", "").replace(" ", "").strip()
    s = s.replace("\xa0", "").replace("\u202f", "")

    try:
        return float(s)
    except ValueError:
        return None

print(clean_price("99000 ₸"))
print(clean_price("25 000"))
print(clean_price("12000.0"))
print(clean_price("abc"))
```
